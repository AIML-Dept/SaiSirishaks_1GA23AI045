```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# Create a 1-qubit circuit
qc = QuantumCircuit(1)

# Apply X gate
qc.x(0)

# Display circuit
print(qc.draw())

# Get statevector
state = Statevector.from_instruction(qc)

print("Statevector:")
print(state)
fig = plot_bloch_multivector(state)
display(fig)
```

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
import matplotlib.pyplot as plt

qc = QuantumCircuit(1)

qc.h(0)
qc.s(0)

print(qc.draw())

state = Statevector.from_instruction(qc)

print(state)

plot_bloch_multivector(state)
plt.show()
fig = plot_bloch_multivector(state)
display(fig)
```

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc = QuantumCircuit(2)

qc.h(0)
qc.h(1)

print(qc.draw())

state = Statevector.from_instruction(qc)

print("Statevector:")
print(state)

print("\nProbability amplitudes:")

probs = state.probabilities()

for basis, p in zip(["00","01","10","11"], probs):
    print(f"{basis} : {p}")

print("\nSum of probabilities =", sum(probs))
```

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
import matplotlib.pyplot as plt
import numpy as np

theta = np.random.uniform(0, np.pi)

qc = QuantumCircuit(1)

qc.h(0)

# Random phase
qc.p(theta,0)

print(qc.draw())

state = Statevector.from_instruction(qc)

print("Random phase angle =",theta)

plot_bloch_multivector(state)

plt.show()
fig = plot_bloch_multivector(state)
display(fig)
```

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
import matplotlib.pyplot as plt
import numpy as np

theta = np.pi/3
phi = np.pi/4

qc = QuantumCircuit(1)

qc.ry(theta,0)
qc.rz(phi,0)

print(qc.draw())

state = Statevector.from_instruction(qc)

print("Statevector:")
print(state)

plot_bloch_multivector(state)

plt.show()
```
