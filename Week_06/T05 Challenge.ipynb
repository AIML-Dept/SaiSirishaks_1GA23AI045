```python
# Challenge: Generalised Simon's Algorithm
# Accepts ANY secret bitstring
# Automatically handles the all-zero edge case

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator


def simons_algorithm(secret, shots=2048):

    
    if not isinstance(secret, str):
        raise ValueError("Secret must be a string.")

    if len(secret) == 0:
        raise ValueError("Secret cannot be empty.")

    if any(bit not in "01" for bit in secret):
        raise ValueError("Secret must contain only 0 and 1.")

    n = len(secret)

    
    if secret == "0" * n:
        print("All-zero edge case detected.")
        print("Secret =", secret)
        print("The oracle is constant.")
        return secret

    
    oracle = QuantumCircuit(n + 1)

    for i, bit in enumerate(secret):
        if bit == "1":
            oracle.cx(i, n)


    qc = QuantumCircuit(n + 1, n)

    
    qc.h(range(n))


    qc.compose(
        oracle,
        qubits=range(n + 1),
        inplace=True
    )


    qc.h(range(n))


    qc.measure(range(n), range(n))

   
    simulator = AerSimulator()

    result = simulator.run(
        qc,
        shots=shots
    ).result()

    counts = result.get_counts()


    print("================================")
    print("SIMON'S ALGORITHM")
    print("================================")
    print("Secret string:", secret)
    print("Number of qubits:", n + 1)

    print("\nCircuit:")
    print(qc.draw())

    print("\nMeasurement results:")
    print(counts)

    return counts


secret = "0000"

results = simons_algorithm(secret)


```
