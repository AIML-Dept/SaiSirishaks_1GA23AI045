```python
import matplotlib.pyplot as plt

n_values = list(range(2, 11))


simon_queries = [
    n for n in n_values
]


classical_queries = [
    2**n for n in n_values
]

plt.figure(figsize=(9, 5))

plt.plot(
    n_values,
    simon_queries,
    marker="o",
    label="Simon's Algorithm"
)

plt.plot(
    n_values,
    classical_queries,
    marker="o",
    label="Classical Brute Force"
)

plt.xlabel("Secret String Length (n)")
plt.ylabel("Number of Oracle Queries")
plt.title("Simon's Algorithm vs Classical Brute Force")

plt.yscale("log")

plt.grid(True)
plt.legend()

plt.show()
```
