```python
from qiskit import QuantumCircuit


secret = "110"
n = len(secret)

def create_simon_oracle(secret):
    n = len(secret)

  
    oracle = QuantumCircuit(n + 1, name="Oracle")

    output = n

    for i, bit in enumerate(secret):
        if bit == "1":
            oracle.cx(i, output)

    return oracle


oracle = create_simon_oracle(secret)

print("Simon Oracle for secret =", secret)
print(oracle.draw())
```

```python
print("Truth Table")
print("-" * 30)
print("Input x   |   f(x)")
print("-" * 30)

for x in range(2**n):
    x_bits = format(x, f"0{n}b")

    result = 0

    for i in range(n):
        if secret[i] == "1":
            result ^= int(x_bits[i])

    print(f"{x_bits}     |   {result}")
```

```python
print("\nChecking Simon's promise f(x) = f(x XOR s)")
print("-" * 45)

valid = True

for x in range(2**n):
    x_bits = format(x, f"0{n}b")

   
    y = int(x_bits, 2) ^ int(secret, 2)
    y_bits = format(y, f"0{n}b")

    def f(bits):
        result = 0
        for i in range(n):
            if secret[i] == "1":
                result ^= int(bits[i])
        return result

    if f(x_bits) != f(y_bits):
        valid = False
        print("Failed for:", x_bits)

if valid:
    print("Oracle is correct!")
    print("f(x) = f(x XOR s) is satisfied.")
```
