```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

def classifier_oracle(label):
    """
    label = 0 -> Class A
    label = 1 -> Class B
    """

    oracle = QuantumCircuit(2)

    if label == 1:
        # Apply phase flip for Class B
        oracle.z(0)

    return oracle

qc = QuantumCircuit(2, 1)
qc.h(0)
qc.x(1)
qc.h(1)
oracle = classifier_oracle(1)

qc.compose(oracle, inplace=True)
qc.h(0)
qc.measure(0, 0)

print(qc)
sim = AerSimulator()

result = sim.run(qc, shots=1024).result()

counts = result.get_counts()

print("Classification Result:", counts)

if '0' in counts:
    print("Predicted Class: A")
else:
    print("Predicted Class: B")
```
