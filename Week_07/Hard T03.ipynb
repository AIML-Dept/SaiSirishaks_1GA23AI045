```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

def create_oracle(function_type):
    """
    Creates an oracle based on the given function type.

    function_type:
        'constant_0' -> f(x) = 0
        'constant_1' -> f(x) = 1
        'balanced'   -> f(x) = x
    """

    oracle = QuantumCircuit(2)

    if function_type == "constant_0":
        # f(x) = 0
        # No gate required
        pass

    elif function_type == "constant_1":
        # f(x) = 1
        oracle.x(1)

    elif function_type == "balanced":
        # f(x) = x
        oracle.cx(0, 1)

    else:
        raise ValueError(
            "Invalid function type. Use "
            "'constant_0', 'constant_1', or 'balanced'."
        )

    return oracle

def deutsch_algorithm(function_type):

    qc = QuantumCircuit(2, 1)
    qc.h(0)
    qc.x(1)
    qc.h(1)
    oracle = create_oracle(function_type)
    qc.compose(oracle, inplace=True)
    qc.h(0)

    qc.measure(0, 0)

    return qc
function_type = "balanced"

qc = deutsch_algorithm(function_type)
print("Deutsch Algorithm Circuit:")
print(qc)
simulator = AerSimulator()

result = simulator.run(
    qc,
    shots=1024
).result()

counts = result.get_counts()

print("\nFunction Type:", function_type)
print("Measurement Results:", counts)

if "0" in counts:
    print("Result: Constant Function")
else:
    print("Result: Balanced Function")
```
