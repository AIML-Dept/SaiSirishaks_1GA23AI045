```python

import numpy as np
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

from qiskit.quantum_info import (
    Statevector,
    DensityMatrix,
    partial_trace
)
simulator = AerSimulator()
print("=" * 60)
print("3-QUBIT GHZ STATE")
print("=" * 60)

ghz = QuantumCircuit(3, 3)

ghz.h(0)
ghz.cx(0, 1)
ghz.cx(1, 2)
ghz.measure([0, 1, 2], [0, 1, 2])

print("\nGHZ Circuit:")
print(ghz)
compiled = transpile(ghz, simulator)

result = simulator.run(
    compiled,
    shots=1024
).result()

ghz_counts = result.get_counts()

print("\nGHZ Measurement Results:")
print(ghz_counts)

print("""
Expected:

000 ≈ 50%
111 ≈ 50%

All other outcomes ≈ 0%
""")
ghz_no_measurement = QuantumCircuit(3)

ghz_no_measurement.h(0)
ghz_no_measurement.cx(0, 1)
ghz_no_measurement.cx(1, 2)

ghz_state = Statevector.from_instruction(
    ghz_no_measurement
)

print("\nGHZ Statevector:")
print(ghz_state)
ghz_density = DensityMatrix(ghz_state)
ghz_reduced = partial_trace(
    ghz_density,
    [0]
)

print("\nGHZ reduced density matrix:")
print(ghz_reduced)
print("\n" + "=" * 60)
print("3-QUBIT W STATE")
print("=" * 60)
w_amplitudes = np.zeros(
    8,
    dtype=complex
)

w_amplitudes[1] = 1 / np.sqrt(3)  # |001>
w_amplitudes[2] = 1 / np.sqrt(3)  # |010>
w_amplitudes[4] = 1 / np.sqrt(3)  # |100>
w_state = Statevector(w_amplitudes)

print("\nW Statevector:")
print(w_state)
w = QuantumCircuit(3, 3)
w.initialize(
    w_amplitudes,
    [0, 1, 2]
)

w.measure(
    [0, 1, 2],
    [0, 1, 2]
)

print("\nW Circuit:")
print(w)
compiled = transpile(w, simulator)

result = simulator.run(
    compiled,
    shots=1024
).result()

w_counts = result.get_counts()

print("\nW Measurement Results:")
print(w_counts)

print("""
Expected:

001 ≈ 33.3%
010 ≈ 33.3%
100 ≈ 33.3%

All other outcomes ≈ 0%
""")
w_density = DensityMatrix(w_state)

# Trace out qubit 0
w_reduced = partial_trace(
    w_density,
    [0]
)

print("\nW reduced density matrix:")
print(w_reduced)
print("\n" + "=" * 60)
print("GHZ vs W COMPARISON")
print("=" * 60)

print("""
GHZ:

|GHZ> = (|000> + |111>) / sqrt(2)

Measurement results:
    000 and 111

If one qubit is measured, the other two become
correlated in the computational basis.

If one qubit is traced out, the remaining two-qubit
state is a classical mixture of |00> and |11>.


W:

|W> = (|001> + |010> + |100>) / sqrt(3)

Measurement results:
    001, 010 and 100

The W state distributes its single excitation
among all three qubits.

If one qubit is lost, entanglement can remain
between the other two qubits.

Therefore GHZ and W states have different
multipartite entanglement structures.
""")
```
