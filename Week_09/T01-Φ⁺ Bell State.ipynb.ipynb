```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

simulator = AerSimulator()

qc = QuantumCircuit(2, 2)

qc.h(0)
qc.cx(0, 1)
qc.measure([0, 1], [0, 1])
print("Phi+ Bell State Circuit:")
print(qc)

compiled_circuit = transpile(qc, simulator)

result = simulator.run(
    compiled_circuit,
    shots=1024
).result()

counts = result.get_counts()

print("\nMeasurement results:")
print(counts)

print("\nChecking outcomes:")

if all(outcome in ["00", "11"] for outcome in counts):
    print("SUCCESS: Only 00 and 11 occurred.")
else:
    print("Unexpected outcome detected.")

print("\nExpected:")
print("00 ≈ 50%")
print("11 ≈ 50%")
```
