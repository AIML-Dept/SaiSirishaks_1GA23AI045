```python
import numpy as np
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

simulator = AerSimulator()

number_of_pairs = 20

alice_results = []
bob_results = []

for i in range(number_of_pairs):

    qc = QuantumCircuit(2, 2)

    # Create Phi+ state
    qc.h(0)
    qc.cx(0, 1)

    # Measure
    qc.measure([0, 1], [0, 1])

    compiled = transpile(qc, simulator)

    result = simulator.run(
        compiled,
        shots=1
    ).result()

    counts = result.get_counts()

    outcome = list(counts.keys())[0]

    # Qiskit returns bit string as c1c0
    bob_bit = int(outcome[0])
    alice_bit = int(outcome[1])

    alice_results.append(alice_bit)
    bob_results.append(bob_bit)

print("=" * 50)
print("QKD / BELL CORRELATION DEMONSTRATION")
print("=" * 50)

print("\nAlice's results:")
print(alice_results)

print("\nBob's results:")
print(bob_results)

matches = 0

for a, b in zip(alice_results, bob_results):

    if a == b:
        matches += 1

print("\nMatching results:",
      matches,
      "out of",
      number_of_pairs)

print(
    "Correlation percentage:",
    matches / number_of_pairs * 100,
    "%"
)


print("""
QKD CONCEPT:

1. Alice and Bob share entangled quantum states.

2. They independently measure their qubits.

3. For compatible measurement bases, their results
   are strongly correlated.

4. They publicly compare selected information about
   their measurement bases.

5. They can compare a subset of results to estimate
   the error rate.

6. Unexpected disturbances can reveal interference
   with the quantum channel.

7. Classical error correction and privacy amplification
   are then used to obtain a secure shared key.

Important:
Entanglement is a resource used by some QKD protocols;
the complete security procedure also requires classical
post-processing.
""")
```
