```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import Aer
from qiskit_aer.noise import NoiseModel, depolarizing_error


qc = QuantumCircuit(1,1)

qc.x(0)
qc.h(0)

qc.measure(0,0)

backend = Aer.get_backend("aer_simulator")

ideal_result = backend.run(
    transpile(qc, backend),
    shots=2048
).result()

print("Ideal Simulator:")
print(ideal_result.get_counts())


noise_model = NoiseModel()

error = depolarizing_error(0.10, 1)

noise_model.add_all_qubit_quantum_error(error, ['h', 'x'])


noisy_result = backend.run(
    transpile(qc, backend),
    shots=2048,
    noise_model=noise_model
).result()

print("\nNoisy Simulator:")
print(noisy_result.get_counts())
```
