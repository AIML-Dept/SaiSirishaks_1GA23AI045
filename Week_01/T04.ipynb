```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import Aer
import random

backend = Aer.get_backend("aer_simulator")

def quantum_random_bits(n_bits=16):

    bits = ""

    for _ in range(n_bits):

        qc = QuantumCircuit(1,1)

        qc.h(0)
        qc.s(0)

        qc.measure(0,0)

        tqc = transpile(qc, backend)

        result = backend.run(tqc, shots=1).result()

        bit = list(result.get_counts().keys())[0]

        bits += bit

    return bits

quantum_bits = quantum_random_bits(16)

classical_bits = ''.join(str(random.randint(0,1)) for _ in range(16))

print("Quantum Random Bits :", quantum_bits)
print("Classical Random Bits:", classical_bits)

print("Quantum 1s :", quantum_bits.count("1"))
print("Classical 1s:", classical_bits.count("1"))
```
