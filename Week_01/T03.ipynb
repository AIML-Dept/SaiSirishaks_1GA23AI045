```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import Aer
import matplotlib.pyplot as plt

def quantum_random_bits(n_bits=30):
    backend = Aer.get_backend('aer_simulator')
    bits = ""

    for _ in range(n_bits):
        qc = QuantumCircuit(1, 1)
        qc.h(0)
        qc.measure(0, 0)

        tqc = transpile(qc, backend)
        result = backend.run(tqc, shots=1).result()

        bits += list(result.get_counts().keys())[0]

    return bits

quantum_bits = quantum_random_bits(30)


zeros = quantum_bits.count('0')
ones = quantum_bits.count('1')

print("Quantum Random Bits :", quantum_bits)
print("Number of 0s :", zeros)
print("Number of 1s :", ones)


plt.figure(figsize=(6,4))
plt.bar(['0', '1'], [zeros, ones])
plt.title("Quantum Random Bits Distribution")
plt.xlabel("Bit Value")
plt.ylabel("Count")
plt.show()
```
