```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
import numpy as np
```

```python

qc = QuantumCircuit(2)

theta = np.pi / 3

qc.ry(theta, 0)
qc.cx(0, 1)


print(qc.draw())


state = Statevector.from_instruction(qc)

print("Statevector:")
print(state)

print("\nProbabilities:")
print(state.probabilities_dict())
```
