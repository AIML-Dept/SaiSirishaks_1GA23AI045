```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
```

```python
# 5 qubits:
# q0 = A
# q1 = B
# q2 = Cin
# q3 = Sum
# q4 = Carry
qc = QuantumCircuit(5, 2)

# Example input:
# A = 1
# B = 1
# Cin = 1

qc.x(0)
qc.x(1)
qc.x(2)


qc.cx(0, 3)
qc.cx(1, 3)
qc.cx(2, 3)


qc.ccx(0, 1, 4)
qc.ccx(0, 2, 4)
qc.ccx(1, 2, 4)


qc.measure(3, 0)
qc.measure(4, 1)

print(qc.draw())


simulator = AerSimulator()

result = simulator.run(qc, shots=1024).result()

counts = result.get_counts()

print("Full Adder Output:")
print(counts)
```
