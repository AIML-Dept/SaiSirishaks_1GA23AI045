```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
import math

simulator = AerSimulator()

# Unknown angle used for the experiment
true_theta = math.radians(60)

shots = 10000

qc = QuantumCircuit(1, 1)

# Apply unknown rotation
qc.ry(true_theta, 0)

# Measure
qc.measure(0, 0)

result = simulator.run(qc, shots=shots).result()

counts = result.get_counts()

ones = counts.get('1', 0)

p1 = ones / shots

# Estimate theta
estimated_theta = 2 * math.asin(math.sqrt(p1))

estimated_degrees = math.degrees(estimated_theta)

print("Counts:", counts)
print("Estimated P(1):", p1)
print("Estimated angle:", estimated_degrees, "degrees")
print("Actual angle:", math.degrees(true_theta), "degrees")
```
