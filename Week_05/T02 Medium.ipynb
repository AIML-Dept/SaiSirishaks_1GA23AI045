```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

simulator = AerSimulator()

qc_plus = QuantumCircuit(1, 1)

# Prepare |+>
qc_plus.h(0)

# Convert X-basis measurement to computational basis
qc_plus.h(0)

# Measure
qc_plus.measure(0, 0)

result_plus = simulator.run(qc_plus, shots=1000).result()

print("Measurement of |+> in X-basis:")
print(result_plus.get_counts())
```

```python
qc_minus = QuantumCircuit(1, 1)

# Prepare |->
qc_minus.x(0)
qc_minus.h(0)

# Measure in X-basis
qc_minus.h(0)

qc_minus.measure(0, 0)

result_minus = simulator.run(qc_minus, shots=1000).result()

print("Measurement of |-> in X-basis:")
print(result_minus.get_counts())
```
