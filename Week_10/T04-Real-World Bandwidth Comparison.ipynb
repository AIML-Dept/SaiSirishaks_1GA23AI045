```python
import pandas as pd
import matplotlib.pyplot as plt

number_of_messages = 100
bits_per_message = 2
classical_bits = number_of_messages * bits_per_message

transmitted_qubits = number_of_messages
data = {
    "Communication Method": [
        "Classical Communication",
        "Superdense Coding"
    ],
    "Information Transmitted": [
        classical_bits,
        classical_bits
    ],
    "Transmitted Units": [
        classical_bits,
        transmitted_qubits
    ],
    "Unit": [
        "Classical bits",
        "Qubits"
    ]
}

df = pd.DataFrame(data)

print(df)
plt.figure(figsize=(8, 5))

plt.bar(
    df["Communication Method"],
    df["Transmitted Units"]
)

plt.ylabel("Number of Transmitted Units")
plt.title("Classical Communication vs Superdense Coding")

plt.show()
```
